# Taller Práctico: Análisis Automatizado de Logs con Python y Pandas

Este cuaderno demuestra cómo procesar registros de sistemas para extraer métricas de seguridad y rendimiento.

```python
import pandas as pd
import matplotlib.pyplot as plt

# Simulación de datos de log de servidor web
data = {
    'ip': ['192.168.1.10', '10.0.0.5', '192.168.1.10', '172.16.0.2', '10.0.0.5', '10.0.0.5'],
    'status': [200, 404, 200, 500, 403, 403],
    'path': ['/index.html', '/admin', '/about', '/api/data', '/wp-login.php', '/wp-login.php']
}

df = pd.DataFrame(data)
print("Resumen de accesos:")
print(df)
```

```python
# Conteo de códigos de estado HTTP
status_counts = df['status'].value_counts()
print(status_counts)

# Gráfica simple de estados
status_counts.plot(kind='bar', color=['green', 'orange', 'red'])
plt.title('Distribución de Códigos HTTP')
plt.xlabel('Código HTTP')
plt.ylabel('Frecuencia')
plt.show()
```
